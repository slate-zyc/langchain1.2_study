# tool_choice的使用

## 1、tool_choice的取值为none

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

# 从.env文件中加载环境变量
load_dotenv(override=True)

# CLOSEAI_API_KEY = os.getenv("CLOSEAI_API_KEY")
# CLOSEAI_BASE_URL = os.getenv("CLOSEAI_BASE_URL")
#
# model = init_chat_model(
#     model="gpt-5.4-mini",
#     model_provider="openai",
#     api_key=CLOSEAI_API_KEY,
#     base_url=CLOSEAI_BASE_URL
# )

CLOSEAI_API_KEY = os.getenv("DEEPSEEK_API_KEY")
CLOSEAI_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=CLOSEAI_API_KEY,
    base_url=CLOSEAI_BASE_URL
)
```

```python
from langchain.tools import tool
from rich import print as rprint

@tool(parse_docstring=True)
def get_weather(city: str) -> str:
    """
    获取当日天气

    Args:
        city: 城市名称
    """
    return f'{city}当天晴朗'


model_with_tools = model.bind_tools([get_weather],tool_choice="none")       # 不调用工具

response = model_with_tools.invoke("北京今天的天气如何？")
rprint(response)
```

## 3、tool_choice取值为required

注意：当模型开启“思考模式”时，API 不允许使用“强制工具调用”。

许多大语言模型（如 DeepSeek、Anthropic 等）的“思考模式”（Thinking Mode）旨在让模型在回答前进行内部推理。为保证推理的开放性和灵活性，API 会禁止同时使用强制性的 tool_choice 参数（如 required、any 或指定具体工具名）

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

# 从.env文件中加载环境变量
load_dotenv(override=True)

CLOSEAI_API_KEY = os.getenv("DEEPSEEK_API_KEY")
CLOSEAI_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    reasoning_effort="none",  # 关闭思考模式
    # extra_body={"thinking": {"type": "disabled"}},  # 第二种写法关闭思考模式
    api_key=CLOSEAI_API_KEY,
    base_url=CLOSEAI_BASE_URL
)

from langchain.tools import tool
from rich import print as rprint

@tool(parse_docstring=True)
def get_weather(city: str) -> str:
    """
    获取当日天气

    Args:
        city: 城市名称
    """
    return f'{city}当天晴朗'


model_with_tools = model.bind_tools([get_weather],tool_choice="required")

response = model_with_tools.invoke("北京今天的天气如何？")
rprint(response)
```
